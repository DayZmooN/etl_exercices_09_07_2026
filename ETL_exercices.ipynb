### Ex E5 ⭐⭐ — Fonction extract()
Encapsulez tout dans une fonction `extract(q1_path, q2_path, clients_path, produits_path) -> dict`
qui retourne un dictionnaire avec les 4 DataFrames harmonisés et consolidés.
La fonction doit afficher des logs à chaque étape.

```python
import logging
logging.basicConfig(level=logging.INFO, format="%(asctime)s | %(message)s", datefmt="%H:%M:%S")

logger = logging.getLogger("ETL")
```

```python
# TON CODE ICI
 

def extract(q1_path, q2_path, clients_path, produits_path) -> dict:
    print("début de l'extraction")
    df_q1 = pd.read_csv(q1_path)
    df_q1["source"] = "q1_2024"
    df_q2 = pd.read_csv(q2_path)
    df_q1["source"] = "q2_2024"

    df_clients = pd.read_csv(clients_path)
    df_produits = pd.read_csv(produits_path)

    print("harmoniser les colonnes")
    df_q2.rename(columns={"moyen_paiement": "mode_paiement"}, inplace=True)

    print("fusion des commandes")
    #librairie loggin
    all_commandes =pd.concat([df_q1,df_q2],ignore_index=True)

    print("remplacemant des devise manque par EUR")
    all_commandes["devise"] = all_commandes["devise"].fillna("EUR")
    display(all_commandes)
    

    print("Extraction terminer")
    


    print("retour d'un tableau Harmoniser et consolidés")
    return [all_commandes, df_clients, df_produits]
    


extract("data/commandes_Q1.csv", "data/commandes_Q2.csv", "data/clients.csv", "data/produits.csv")
```

### Ex T7 ⭐⭐ — Feature Engineering
Créez les colonnes suivantes :
1. `tranche_montant` : `<100€` / `100-500€` / `500-2000€` / `>2000€` avec `pd.cut()`
2. `saison` : depuis le mois (Hiver/Printemps/Été/Automne)
3. `est_annulee` : booléen True si statut == 'annulée'
4. `ca_perdu` : montant_total si commande annulée, sinon 0
5. `delai_inscription_commande` : nb de jours entre date_inscription du client et date_commande

Affichez `value_counts()` pour les colonnes catégorielles créées.

```python
# TON CODE ICI
#1. tranche_montant
df_resultat_cmd_clt["tranche_montant"] = pd.cut(res_q1_q2["montant_total"],  bins=[-float("inf"),100, 500,2000, float("inf")], labels=[  "<100€",
    "100-500€",
    "500-2000€",
    ">2000€"])
display(df_resultat_cmd_clt["tranche_montant"].value_counts())

# help(pd.cut)
#2. saison 
def saison(mois):
    if  mois in [12,1,2]:
        return 'hiver'
    elif 3 <= mois <= 5:
        return 'printemps'
    elif 5 < mois <= 8:
        return "été"
    elif 8 < mois <= 11:
        return "automne"
    else :
        return "erreur de valeur"

df_resultat_cmd_clt["saison"] = df_resultat_cmd_clt["mois"].apply(saison)
df_resultat_cmd_clt.head(4)

#3.est annulee
df_resultat_cmd_clt["statut"].head()
df_resultat_cmd_clt["est_annulee"] =df_resultat_cmd_clt["statut"] == "annulée"
#4. ca_perdu
df_resultat_cmd_clt["ca_perdu"] = 0.0
df_resultat_cmd_clt.loc[df_resultat_cmd_clt["est_annulee"]==True,["ca_perdu"]] = df_resultat_cmd_clt["montant_total"]
df_resultat_cmd_clt.head(2)

#5. delai inscription
df_resultat_cmd_clt[["date_commande","date_inscription"]].dtypes
df_resultat_cmd_clt["date_inscription"] = pd.to_datetime(df_resultat_cmd_clt["date_inscription"], format="mixed")
df_resultat_cmd_clt["delai_inscription_commande"] = df_resultat_cmd_clt["date_commande"] - df_resultat_cmd_clt["date_inscription"]
df_resultat_cmd_clt["saison"] = df_resultat_cmd_clt["mois"].apply(saison)

#afficher les  value_count() des cateories
print(df_resultat_cmd_clt["tranche_montant"].value_counts())
print(df_resultat_cmd_clt["saison"].value_counts())
print(df_resultat_cmd_clt["est_annulee"].value_counts())
print(df_resultat_cmd_clt["ca_perdu"].value_counts())
print(df_resultat_cmd_clt["delai_inscription_commande"].value_counts())
```

### Ex T8 ⭐⭐⭐ — Validation qualité
Écris une fonction `valider_pipeline(df) -> bool` qui :
1. Vérifie qu'il n'y a plus de NaN sur les colonnes critiques
2. Vérifie qu'il n'y a plus de doublons
3. Vérifie que tous les prix sont > 0
4. Vérifie que toutes les quantités sont > 0
5. Vérifie que montant_total == quantite × prix_unitaire (à 1 centime près)
6. Retourne True si tout est OK, False sinon avec les erreurs détaillées

```python
# TON CODE ICI
def valider_pipeline(df) -> bool:
    #1. vérifie plus de nan sur les colonnes critiques
    erreur = []
    df_critique_nan = df[["client_id","commande_id","produit_id","montant_total","quantite","date_commande", "prix_unitaire",]].isna().sum()

    for  colonne, valeur in df_critique_nan.items():
        if valeur > 0:
            erreur.append(f"{colonne}: {valeur}")
    
    #2. doublons
    s = df.duplicated().sum()
    print(s)
    if s > 0:
        erreur.append(s)
        display(f"il y' a {erreur} de doulons")
    
    #3. prix > 0
    df_supp_zero = df["prix_unitaire"].sum()
    if df_supp_zero <= 0 :
        print(f"prix inferieur a zero : {df["prix_unitaire"]}")
        erreur.append(f"prix inferreur a zero {df_supp_zero}")
    elif df_supp_zero > 0:
        print("aucun prix inferieur a Zéro")
    else:
        print("une erreur de prix 'valeur' non conforme")

    #4. quantite > 0 
    df_supp_quantite  = df["quantite"].sum()
    if df_supp_quantite <= 0 :
        print(f"quantite inferieur a zero : {df["quantite"]}")
        erreur.append(f"quantite inferreur a zero {df_supp_quantite}")
    elif df_supp_quantite > 0:
        print("aucun quantite inferieur a Zéro")
    else:
        print("une erreur de quantite 'valeur' non conforme")

    #5. verifiez montant_total = quantite * prix_unitaire
    df_cal_prix = df["prix_unitaire"].round(2) * df["quantite"]
    ecart = df["montant_total"].round(2) - df_cal_prix.round(2)
    if ((ecart > 0.01) | (ecart < -0.01)).any():
        erreur.append(f"le prix et superieur ou inferieur a {ecart}")
        print("le prix est inferieur au critere ou supperieur au centimes pret")
    else:
        print("le prix corespond au centimes pret")
    
    if erreur  :
        print(f"eurreur {erreur}")
        return False

    logger.info("Pipline valine aucune erreur a etait detecter")
    return True 

valider_pipeline(df_resultat_cmd_clt)
```
